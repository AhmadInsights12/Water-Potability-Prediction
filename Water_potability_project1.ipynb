```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('water_potability.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df_water = df.fillna(df.mean())
```

```python
df_water
```

```python
list(df_water.columns)
```

```python
col_list = list(df_water.columns)
```

```python
col_list [0 : -1]
```

```python
#ya  humara pass input hn ess liya hum es ko X ma store krta DisplayHandle

x = df_water[col_list [0 : -1]]
```

```python
y = df_water['Potability']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
lg_model = LogisticRegression()
lg_model.fit(x_train, y_train)

# hum na ess algorithm pa check kiya hn model ko lkn es pa model siee km ni kr raha ta 
```

```python
pred_lg = lg_model.predict(x_test)
```

```python
pred_lg
```

```python
from sklearn.metrics import accuracy_score
acc = round(accuracy_score(y_test, pred_lg) * 100, 2)
print(f'Accuracy:{acc} %')
```

```python
# ab ya model sieee km ni kra raha 
# ess sa lagta ha ka ya lagorithm siee km ni kr raha hn
```

```python
from sklearn.tree import DecisionTreeClassifier
dc_model = DecisionTreeClassifier()
dc_model.fit(x_train, y_train)
```

```python
pred_dc = dc_model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
acc = round(accuracy_score(y_test, pred_dc) * 100, 2)
print(f'Accuracy:{acc} %')
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_model = RandomForestClassifier()
rf_model.fit(x_train, y_train)
```

```python
pred_rf = rf_model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
acc = round(accuracy_score(y_test, pred_rf) * 100, 2)
print(f'accuracy: {acc} %')
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
sns.heatmap(confusion_matrix(y_test, pred_rf), annot= True, cmap = 'viridis', fmt='.1f')
```

```python
y.value_counts()
```

```python
from imblearn.under_sampling import RandomUnderSampler
from imblearn.over_sampling import RandomOverSampler
```

```python
under = RandomUnderSampler(random_state=42)

x_under, y_under = under.fit_resample(x_train, y_train)

y_under.value_counts()
```

```python
rf_under = RandomForestClassifier()

rf_under.fit(x_under, y_under)

pred_under = rf_under.predict(x_test)

acc_under = accuracy_score(y_test, pred_under) * 100

print(f'Under Sampling Accuracy: {acc_under:.2f}%')
```

```python
over = RandomOverSampler(random_state=42)

x_over, y_over = over.fit_resample(x_train, y_train)

y_over.value_counts()
```

```python
rf_over = RandomForestClassifier()

rf_over.fit(x_over, y_over)

pred_over = rf_over.predict(x_test)

acc_over = accuracy_score(y_test, pred_over) * 100

print(f'Over Sampling Accuracy: {acc_over:.2f}%')
```

```python
print(f'Original Accuracy: {acc:.2f}%')
print(f'Under Sampling: {acc_under:.2f}%')
print(f'Over Sampling: {acc_over:.2f}%')
```

```python
import joblib

model_data = {
    'model': rf_under,
    'features': list(x_train.columns),
    'defaults': x_train.median().to_dict()
}

joblib.dump(model_data, 'water_model.pkl')
```
